# Debug Null Bytes in booking_service.py

Inspect the raw bytes of `backend/services/booking_service.py` to identify embedded null bytes or encoding issues.

```python
from pathlib import Path
p = Path('services/booking_service.py')
raw = p.read_bytes()
if raw.endswith(b'\x00'):
    p.write_bytes(raw[:-1])
    print('trimmed null byte')
else:
    print('no trimming needed')
print('new length', len(p.read_bytes()))
print('endswith null', p.read_bytes().endswith(b'\x00'))
```
